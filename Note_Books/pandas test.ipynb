```python
import pandas as pd

pd.set_option("display.float_format", "{:,.2f}".format)

df = pd.read_excel("d:/Users/1443/Downloads/CFMS_Recon_25-26.xlsx")

result = df.groupby("FLAG")["AMOUNT"].sum()

print(result["P"])
print(result["R"])
print(result["P"] == result["R"])
print(result["P"] - result["R"])
```
